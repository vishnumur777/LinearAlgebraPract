## **Scalar and Vector Projection**

**Scalar Projection:** The scalar projection of a vector a onto another vector b (denoted  'proj<sub>b</sub>a') is a measure of how much of a lies in the direction of b, but only as a scalar value (magnitude, without direction).

**Formula for scalar projection:** a.b / |b|

Where,

    • a.b = dot product of 'a' and 'b' vectors.
    
    • |b| = magnitude of 'b' vectors.

**Vector Projection:** The vector projection of a vector a onto another vector b is a vector that represents the component of a that lies along the direction of b. This projection has both magnitude and direction.

**Formula for vector projection:** ( a.b / b.b )b (or)  a.b / ||b||<sup>2</sub>

Where,

  • a.b = dot product of 'a' and 'b' vectors.

  • ||b||<sup>2</sup> = square of magnitude of 'b' vectors.

## **Linear Combination, Basis, Span, Linear Independence**

### 1. Linear Combination
**Definition:** A linear combination of vectors $v_1, v_2, \ldots, v_n$ is an expression of the form $$ s_1 v_1 + s_2 v_2 + \ldots + s_n v_n $$, where $s_i$ are scalars.

### 2. Linear Independence
**Definition:** A set of vectors $S = \{v_1, v_2, \ldots, v_k\}$ is said to be **linearly independent** if none of the vectors $v_i$ can be written as a linear combination of the other vectors. Equivalently, the only solution to the equation $s_1 v_1 + s_2 v_2 + \ldots + s_k v_k = 0$ is when all $s_i = 0$. If there exists a solution where not all $s_i = 0$, then the vectors are **linearly dependent**.

### 3. Span of Vectors
**Definition:** The **span** of a set of vectors $\{v_1, v_2, \ldots, v_k\}$ is the set of all vectors that can be written as a linear combination of these vectors. It is denoted as $\text{span}\{v_1, v_2, \ldots, v_k\}$ and includes all possible linear combinations of the given vectors.

### 4. Basis
**Definition:** A **basis** for a vector space V is a set of vectors $B = \{b_1, b_2, \ldots, b_m\}$ that satisfies two conditions:
- **Spanning:** The set B spans the vector space V, meaning any vector in V can be written as a linear combination of the vectors in B.
- **Linear Independence:** The set B is linearly independent, meaning none of the vectors in B can be written as a linear combination of the others.

### Example
Consider the vectors $v_1 = (2, 0, 0)$, $v_2 = (0, 1, 0)$, and $v_3 = (0, 0, 7)$. These vectors are linearly independent because none can be written as a linear combination of the others, and the only solution to $s_1 v_1 + s_2 v_2 + s_3 v_3 = 0$ is when $s_1 = s_2 = s_3 = 0$. The span of these vectors is $\mathbb{R}^3$ because any vector in $\mathbb{R}^3$ can be expressed as a linear combination of these vectors. Therefore, these vectors form a basis for $\mathbb{R}^3$.

### **Analogies**

To understand the concepts of linear combinations, linear independence, basis, and span of vectors, we can use an analogy involving a set of building blocks or LEGO pieces.

### 1. **Linear Combinations**
**Analogy:** Imagine you have a set of LEGO pieces of different shapes and colors. A linear combination is like constructing a new structure by using these LEGO pieces, where each piece can be used multiple times and scaled (in this case, by using multiples of the same piece). For example, if you have two types of LEGO pieces, you can build a new structure by using 3 of the first type and 2 of the second type. This is similar to how vectors are combined: $$3\vec{v}_1 + 2\vec{v}_2$$.

### 2. **Linear Independence**
**Analogy:** Now, consider the LEGO pieces again. If you have a set of pieces where no piece can be constructed by combining the others, then these pieces are like linearly independent vectors. For instance, if you have a square piece and a triangular piece, and neither can be made by combining the other, they are linearly independent. If, however, you could make the triangular piece by combining multiple square pieces, then they would not be linearly independent.

### 3. **Basis**
**Analogy:** A basis is like a minimal set of LEGO pieces that can be used to build any structure you want. If you have a set of pieces that are linearly independent and can be combined to build any possible structure, then this set is a basis. For example, if you have a set of basic LEGO shapes (like squares, triangles, and circles) that cannot be made from each other but can be combined to build any structure, these shapes form a basis.

### 4. **Span**
**Analogy:** The span of a set of vectors is like the collection of all possible structures you can build using your LEGO pieces. If you have a set of LEGO pieces, the span is the set of all structures that can be constructed by combining these pieces in any way. For instance, if you have a set of square and triangular pieces, the span would include all structures that can be built using these pieces, such as houses, cars, or any other structure.

### Summary
- **Linear Combinations:** Combining LEGO pieces to build new structures.
- **Linear Independence:** Ensuring no piece can be made from the others.
- **Basis:** A minimal set of pieces that can build any structure.
- **Span:** The collection of all possible structures built from the pieces.

This analogy helps to visualize how these concepts work together in linear algebra.

### **Matrix Transformation**

**Matrix transformation** is a mathematical concept where a matrix is used to represent and perform geometric transformations on vectors. Here’s a detailed overview:

### What is Matrix Transformation?

A matrix transformation is a function that takes a vector in one space and maps it to another vector in the same or a different space using matrix multiplication. This is represented as $$ T(\mathbf{x}) = A\mathbf{x} $$, where $A$ is the transformation matrix, and $\mathbf{x}$ is the input vector.

### Types of Matrix Transformations

Matrix transformations can be categorized into several types based on the geometric operations they perform:

1. **Translation Matrix**:
   - Moves an object along one or more axes. For example, in 3D space, a translation matrix can be represented as:
     $$
     \begin{bmatrix}
     1 & 0 & 0 & T_x \\
     0 & 1 & 0 & T_y \\
     0 & 0 & 1 & T_z \\
     0 & 0 & 0 & 1
     \end{bmatrix}
     \cdot
     \begin{pmatrix}
     x \\ y \\ z \\ 1
     \end{pmatrix}
     =
     \begin{pmatrix}
     x + T_x \\ y + T_y \\ z + T_z \\ 1
     \end{pmatrix}
     $$
.

2. **Rotation Matrix**:
   - Rotates an object around a specific axis. For example, a rotation matrix in 2D space can be:
     $$
     \begin{bmatrix}
     \cos(\theta) & -\sin(\theta) \\
     \sin(\theta) & \cos(\theta)
     \end{bmatrix}
     $$
.

3. **Scaling Matrix**:
   - Scales an object by a factor along one or more axes. For example, a scaling matrix in 2D space can be:
     $$
     \begin{bmatrix}
     k_x & 0 \\
     0 & k_y
     \end{bmatrix}
     $$
.

4. **Reflection Matrix**:
   - Reflects an object across a specific axis or plane. For example, a reflection matrix in 2D space across the x-axis can be:
     $$
     \begin{bmatrix}
     1 & 0 \\
     0 & -1
     \end{bmatrix}
     $$
.

5. **Shear Matrix**:
   - Shears an object, changing its shape by shifting it along one axis in proportion to its distance from another axis. For example, a shear matrix in 2D space can be:
     $$
     \begin{bmatrix}
     1 & b \\
     0 & 1
     \end{bmatrix}
     $$
.

6. **Affine Transformation Matrix**:
   - Preserves lines and parallelism, often used in design software. Affine transformations can include combinations of translation, rotation, scaling, and shear.

7. **Projection Matrix**:
   - Projects an object onto a line or plane. This can be used to reduce the dimensionality of a vector.

### Composition of Transformations

One of the powerful aspects of matrix transformations is the ability to compose multiple transformations by multiplying their respective matrices. The order of multiplication is crucial, as it determines the sequence in which the transformations are applied.

In summary, matrix transformations provide a versatile and efficient way to perform various geometric operations, making them essential tools in fields such as computer graphics, engineering, and physics.

### **Understanding Rotation Matrices**

To rotate a 2D vector $$\vec{x}_1 = [4, 5]$$ by 30 degrees, you can use the rotation matrix formula. Here is the step-by-step process:

1. **Define the Rotation Matrix:**
   The rotation matrix for a 2D rotation by an angle $\theta$ is given by:
   $$
   R(\theta) = \begin{bmatrix}
   \cos \theta & -\sin \theta \\
   \sin \theta & \cos \theta
   \end{bmatrix}
   $$
   For a rotation of 30 degrees, we need to compute $\cos(30^\circ)$ and $\sin(30^\circ)$. These values are:
   $$
   \cos(30^\circ) = \frac{\sqrt{3}}{2} \approx 0.866
   $$
   $$
   \sin(30^\circ) = \frac{1}{2} \approx 0.5
   $$
   So, the rotation matrix for a 30-degree rotation is:
   $$
   R(30^\circ) = \begin{bmatrix}
   0.866 & -0.5 \\
   0.5 & 0.866
   \end{bmatrix}
   $$

2. **Apply the Rotation Matrix:**
   Now, multiply the vector $$\vec{x}_1 = [4, 5]$$ by the rotation matrix:
   $$
   \begin{bmatrix}
   x' \\ y'
   \end{bmatrix} = R(30^\circ) \begin{bmatrix}
   4 \\ 5
   \end{bmatrix}
   $$
   Performing the matrix multiplication:
   $$
   x' = 4 \cdot 0.866 - 5 \cdot 0.5 = 3.464 - 2.5 = 0.964
   $$
   $$
   y' = 4 \cdot 0.5 + 5 \cdot 0.866 = 2 + 4.33 = 6.33
   $$
   Therefore, the rotated vector is:
   $$
   \vec{x}'_1 = [0.964, 6.33]
   $$

### Summary

- **Rotation Matrix:** 
  $$
  R(30^\circ) = \begin{bmatrix}
  0.866 & -0.5 \\
  0.5 & 0.866
  \end{bmatrix}
  $$
- **Rotated Vector:** 
  $$
  \vec{x}'_1 = [0.964, 6.33]
  $$

This method ensures that the vector is rotated correctly by 30 degrees in the counterclockwise direction.

### **Linear Transformation**

### Linear Transformation of Vectors

A **linear transformation** is a function between vector spaces that preserves the operations of vector addition and scalar multiplication. Here is a detailed explanation:

- **Definition**: A linear transformation $T: \mathbb{R}^n \to \mathbb{R}^m$ is a function that maps vectors from an $n$-dimensional vector space to an $m$-dimensional vector space, satisfying two key properties:
  1. **Additivity**: $T(\vec{x} + \vec{y}) = T(\vec{x}) + T(\vec{y})$ for all vectors $\vec{x}$ and $\vec{y}$ in $\mathbb{R}^n$.
  2. **Homogeneity**: $T(k\vec{x}) = kT(\vec{x})$ for all vectors $\vec{x}$ in $\mathbb{R}^n$ and all scalars $k$.

### Types of Linear Transformations with Examples

1. **Matrix Transformations**:
   - These are the most common form of linear transformations, where the transformation is defined by matrix multiplication. For example, given a matrix $A$, the transformation $T: \mathbb{R}^n \to \mathbb{R}^m$ defined by $$T(\vec{x}) = A\vec{x}$$ is a linear transformation.

   **Example**: $$A = \begin{pmatrix} 1 & 2 \\ 2 & 1 \end{pmatrix}$$. Then $T: \mathbb{R}^2 \to \mathbb{R}^2$ defined by $$T\left(\begin{pmatrix} x \\ y \end{pmatrix}\right) = A\begin{pmatrix} x \\ y \end{pmatrix} = \begin{pmatrix} x + 2y \\ 2x + y \end{pmatrix}$$ is a linear transformation.

2. **Identity Transformation**:
   - This transformation maps every vector to itself. It is defined by $T(\vec{x}) = \vec{x}$ for all $\vec{x}$ in the vector space.

   **Example**: $T: \mathbb{R}^2 \to \mathbb{R}^2$ defined by $$T\left(\begin{pmatrix} x \\ y \end{pmatrix}\right) = \begin{pmatrix} x \\ y \end{pmatrix}$$.

3. **Zero Transformation**:
   - This transformation maps every vector to the zero vector. It is defined by $T(\vec{x}) = \vec{0}$ for all $\vec{x}$ in the vector space.

   **Example**: $T: \mathbb{R}^2 \to \mathbb{R}^2$ defined by $$T\left(\begin{pmatrix} x \\ y \end{pmatrix}\right) = \begin{pmatrix} 0 \\ 0 \end{pmatrix}$$.

4. **Projection Transformations**:
   - These transformations project vectors onto a subspace. For example, projecting a vector in $\mathbb{R}^3$ onto the $xy$-plane.

   **Example**: $T: \mathbb{R}^3 \to \mathbb{R}^3$ defined by $$T\left(\begin{pmatrix} x \\ y \\ z \end{pmatrix}\right) = \begin{pmatrix} x \\ y \\ 0 \end{pmatrix}$$.

5. **Rotation Transformations**:
   - These transformations rotate vectors around an axis. For example, rotating a vector in $\mathbb{R}^2$ by an angle $\theta$.

   **Example**: $T: \mathbb{R}^2 \to \mathbb{R}^2$ defined by $$T\left(\begin{pmatrix} x \\ y \end{pmatrix}\right) = \begin{pmatrix} \cos\theta & -\sin\theta \\ \sin\theta & \cos\theta \end{pmatrix} \begin{pmatrix} x \\ y \end{pmatrix}$$.

6. **Scaling Transformations**:
   - These transformations scale vectors by a factor. For example, scaling a vector in $\mathbb{R}^2$ by a factor $k$.

   **Example**: $T: \mathbb{R}^2 \to \mathbb{R}^2$ defined by $$T\left(\begin{pmatrix} x \\ y \end{pmatrix}\right) = \begin{pmatrix} kx \\ ky \end{pmatrix}$$.

7. **Inverse Transformation**
    - An **inverse transformation** is a type of linear transformation that reverses the effect of another linear transformation. For a linear transformation $ T: V \to W $, the inverse transformation $T^{-1}: W \to V $satisfies:
        - $ T^{-1}(T(\vec{v})) = \vec{v} $ for all $ \vec{v} $ in $ V $
        - $ T(T^{-1}(\vec{w})) = \vec{w} $ for all $ \vec{w} $ in $ W $

    For example, if $ T: \mathbb{R}^2 \to \mathbb{R}^2 $ is defined by $ T(x, y) = (2x, 3y) $, then the inverse transformation $ T^{-1} $ would be defined by $$ T^{-1}(x, y) = \left(\frac{x}{2}, \frac{y}{3}\right) $$.


### Verification of Linearity

To verify that a transformation is linear, one must check that it satisfies the two properties mentioned above: additivity and homogeneity. Here is an example of verifying linearity for a transformation:

**Example Verification**:
Consider the transformation $T: \mathbb{R}^2 \to \mathbb{R}^3$ defined by $$T\left(\begin{pmatrix} x \\ y \end{pmatrix}\right) = \begin{pmatrix} x^2 \\ 2x \\ xy \end{pmatrix}$$. To check if this is linear, we need to verify the two properties:

1. **Additivity**: Let $\vec{x} = \begin{pmatrix} x_1 \\ y_1 \end{pmatrix}$ and $\vec{y} = \begin{pmatrix} x_2 \\ y_2 \end{pmatrix}$. Then,
   $$
   T(\vec{x} + \vec{y}) = T\left(\begin{pmatrix} x_1 + x_2 \\ y_1 + y_2 \end{pmatrix}\right) = \begin{pmatrix} (x_1 + x_2)^2 \\ 2(x_1 + x_2) \\ (x_1 + x_2)(y_1 + y_2) \end{pmatrix}
   $$
   This does not equal $T(\vec{x}) + T(\vec{y})$, so the transformation is not linear.

This example illustrates that not all transformations are linear, and verifying linearity is crucial.

The goal is to convert above augumented matrix into **Row echoleon Form**

## _**Row Echelon Form**_

A matrix is in **row echelon form** if it satisfies the following conditions:

## Conditions for Row Echelon Form

1. **Zero Rows at the Bottom**:
   - All rows consisting entirely of zeros must be located at the bottom of the matrix.

2. **Leading Entries**:
   - Each non-zero row must have a **leading entry** (also known as a **pivot**), which is the **left-most non-zero entry in that row** (or) **First non-zero element in that row.** This leading entry must be to the right of the leading entry in the row above it.

3. **Entries Below Leading Entries**:
   - All entries in a column below a leading entry must be zeros. This implies that each leading entry is the only non-zero entry in its column below it.

### Optional Condition on Leading Entries
- Some definitions and texts may not require that the leading entry be 1, while others do. However, this requirement is typically reserved for the **reduced row echelon form**. In general row echelon form, the leading entry can be any non-zero number.

### Examples and Clarifications

- Here is an example of a matrix in row echelon form:
  $$
  \begin{bmatrix}
  1 & 2 & -1 & 4 \\
  0 & 1 & 0 & 3 \\
  0 & 0 & 1 & 2 \\
  0 & 0 & 0 & 0
  \end{bmatrix}
  $$
  This matrix satisfies all the conditions: zero rows are at the bottom, and each leading entry is to the right of the one above it.

- A matrix is not in row echelon form if it violates any of these conditions, such as having a non-zero row below a zero row or having leading entries in the same column.

By ensuring these conditions are met, a matrix can be transformed into row echelon form using Gaussian elimination, which is particularly useful for solving systems of linear equations.

Let's come into augumented matrix $ A $.

Inorder to convert $ A $ into Row Echelon Form, we have to do some matrix row operations.

By applying $ R_3 \rightarrow R_1 + R_3 $, we get

$$
A = \begin{bmatrix}

$$ 

Now the given matrix is in row echelon form, why?

$$
A = \begin{bmatrix}
\begin{array}{ccc|c}
1 & 1 & -1 & -2 \\
0 & -3 & 3 & 9 \\
0 & 0 & 4 & 8 \\
\end{array}
\end{bmatrix}
$$

### Explanation:
1. The leading entry in the first row is 1 (in the first column).
2. The second row's leading entry is \(-3\), which is in the second column, to the right of the leading entry in the first row.
3. The third row's leading entry is 4, which is in the third column, to the right of the leading entry in the second row.
4. No zero rows, and all non-zero rows are above any potential zero rows (though there aren't any here).

Thus, all the conditions for row echelon form are satisfied.

### **Gaussian Elimination Method and Finding Inverse matrix**

- Linear equations can be solved in a few different ways
- Most commonly applied techniques are elimination method and subsitution method.
- In linear algebra, we have a different technique using matrix inversion.
- The concept of inversion is well known starting from real numbers.

- A real number $ A $ has a multiplicative inverse if there exists a number b such that a multiplied with b equals 1.

$$ a.b = 1 \rightarrow A.B = I $$

- We can generalize the concept of mulitplicative inverses to matrices.
- Inverse of matrix A is denoted as $ A^{-1}$.


$$
    A.A^{-1} = I
$$
$$
    A.B = B.A = I
$$

- A matrix B such that $ A.B = B.A = I $ is called the inverse of the matrix A.
- If matrix A is invertible and we want to solve the sytem, then for any vector b the system has a unique solution x.
$$ 
A.x = B
$$
- We can get x by multiplying both sides of an equation of the left by the matrix $ A^{-1} $.


From the equation; $A.x = B$
multiply $A^{-1}$ on both sides. We get,

$$
A^{-1}.A.x = A^{-1}.B
$$

since $A.A^{-1} = I$,

$$
x = A^{-1}.B
$$

```python
import numpy as np
```

```python
A = np.array([[12,32],[23,64]])
A
```

Find inverse of A. ($A^{-1}$)

```python
Ainv = np.linalg.inv(A)
```

```python
Ainv
```

Initialize   $\vec{b}$

```python
b = np.array([41,20])
b
```

Apply $ x = A^{-1}.\vec{b}$, we get

```python
x = np.dot(Ainv,b)
x
```

To verify $\vec{b}$, we can just multiply $A$ and $x$,

```python
np.dot(A,x)
```

### **Gram-Schmidt Process**

- Used to transform any basis to orthogonal basis.

### Steps to do Gram-Schmidt

- If we consider a matrix with 5x5 dimensions.
- The first column will be the same column.
- Take second column orthogonalize relative to first column.
- Apply third column orthogonlize relative to second and first column.
- Apply fourth column orthogonalize relative to 3rd,2nd and 1st column.
- Apply fifth column orthogonalize relative to 4th,3rd,2nd and 1st column.
### 
- Since all the columns in the matrix are orthogonal, but it is not a orthogonal matrix,
- for that, normalize each column in the matrix (mulitplying with magnitude).
- Finally, we get orthogonalized matrix with all columns pairwise with unit length.
